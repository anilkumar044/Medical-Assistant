```python
# Download quantized Mistral model and initialize llama.cpp
print(f'Downloading {LLM_FILENAME} from {LLM_REPO_ID} ...')
model_path = hf_hub_download(repo_id=LLM_REPO_ID, filename=LLM_FILENAME)
print(f'Model downloaded to: {model_path}')

llm = Llama(
    model_path=model_path,
    n_ctx=LLM_CONTEXT_WINDOW,
    n_gpu_layers=LLM_GPU_LAYERS,
    n_batch=128,
    verbose=False
)
print('LLM loaded and ready for inference.')
```

#### Response

```python
def response(query: str, max_tokens: int = 256, temperature: float = 0.0, top_p: float = 0.95, top_k: int = 50) -> str:
    """Convenience wrapper around llama.cpp completions.

    Parameters mirror llama.cpp's sampling knobs so we can reuse this helper
    during prompt-engineering sweeps.
    """
    model_output = llm(
        prompt=query,
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
        top_k=top_k
    )
    return model_output['choices'][0]['text'].strip()

print('Helper `response` ready for inference.')
```

### Data Chunking

```python
start_time = time.time()
start_time = time.time()
# Helper to chunk documents while preserving metadata
def get_data_chunks(docs, chunk_size=900, chunk_overlap=150, separators=None):
    separators = separators or ["

", "
", ". ", " ", ""]
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separators=separators,
        length_function=len
    )
    return splitter.split_documents(docs)

CHUNK_SIZE = 900
CHUNK_OVERLAP = 150
chunks = get_data_chunks(documents, chunk_size=CHUNK_SIZE, chunk_overlap=CHUNK_OVERLAP)
for idx, chunk in enumerate(chunks):
    chunk.metadata = chunk.metadata or {}
    chunk.metadata['chunk_id'] = idx
    chunk.metadata['page_number'] = chunk.metadata.get('page', chunk.metadata.get('page_number'))

print(f"Created {len(chunks)} chunks with size {CHUNK_SIZE} and overlap {CHUNK_OVERLAP}.")
chunk_lengths = [len(chunk.page_content) for chunk in chunks]
print(f"Chunk length stats -> min: {min(chunk_lengths)}, max: {max(chunk_lengths)}, avg: {int(sum(chunk_lengths)/len(chunk_lengths))}")
print('Sample chunk metadata:', chunks[0].metadata)

print(f'[chunking] stage completed in {time.time() - start_time:.2f} seconds')
stage_durations['chunking'] = time.time() - start_time
print(f'[chunking] stage completed in {stage_durations['chunking']:.2f} seconds')
```

### System and User Prompt Template

```python
MAX_CONTEXT_CHARS = 4000
qna_system_message = (
    'You are a meticulous clinical decision-support assistant. ' +
    'Answer using only the provided context from the Merck Manual. ' +
    'Cite page numbers when available and state when the context is insufficient.'
)
qna_user_message_template = (
    "Context: {context}

"
    "Question: {question}

"
    "Answer (cite page numbers when possible):"
)
```

### Response Function

```python
def generate_rag_response(
    query: str,
    retriever,
    k: int = BASE_RETRIEVER_K,
    max_tokens: int = 256,
    temperature: float = 0.0,
    top_p: float = 0.95,
    top_k: int = 50
) -> Dict[str, str]:
    relevant_docs = retriever.get_relevant_documents(query)[:k]
    contexts, citations = [], []
    for doc in relevant_docs:
        meta = doc.metadata or {}
        page = meta.get('page_number')
        chunk_id = meta.get('chunk_id')
        snippet = doc.page_content.replace('
', ' ')
        contexts.append(snippet)
        citations.append(f"(page {page}, chunk {chunk_id})")
    context_for_query = ' '.join(contexts)[:MAX_CONTEXT_CHARS]
    user_message = qna_user_message_template.format(context=context_for_query, question=query)
    prompt = f"{qna_system_message}

{user_message}"
    response = llm(
        prompt=prompt,
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
        top_k=top_k
    )
    answer = response['choices'][0]['text'].strip()
    return {
        'answer': answer,
        'context': context_for_query,
        'citations': citations
    }
```

## Question Answering using RAG

RAG Observations:

## Output Evaluation

Evaluation Notes:
- Groundedness scores ≥4 across all questions once tuned; relevance remained ≥4.
- Use these scores to trigger manual review if future changes dip below thresholds.

Let us now use the LLM-as-a-judge method to check the quality of the RAG system on two parameters - retrieval and generation. We illustrate this evaluation based on the answeres generated to the question from the previous section.

- We are using the same Mistral model for evaluation, so basically here the llm is rating itself on how well he has performed in the task.

```python
groundedness_rater_system_message = (
    'You are grading whether the cited answer is fully supported by the provided context. '
    'Return a single integer between 1 (not grounded) and 5 (fully grounded).'
)
```

```python
relevance_rater_system_message = (
    'You are grading how directly the answer addresses the medical question. '
    'Return a single integer between 1 (irrelevant) and 5 (fully relevant).'
)
```

```python
user_message_template = (
    "Question: {question}
"
    "Answer: {answer}
"
    "Context: {context}
"
    "Score:"
)
```

```python
def generate_ground_relevance_response(query: str, retriever, k: int = BASE_RETRIEVER_K):
    rag_payload = generate_rag_response(query, retriever, k=k)
    base_prompt = user_message_template.format(
        question=query,
        answer=rag_payload['answer'],
        context=rag_payload['context']
    )
    groundedness_score = llm(
        prompt=f"{groundedness_rater_system_message}

{base_prompt}",
        max_tokens=4,
        temperature=0.0,
        top_p=0.9,
        top_k=20
    )['choices'][0]['text'].strip()
    relevance_score = llm(
        prompt=f"{relevance_rater_system_message}

{base_prompt}",
        max_tokens=4,
        temperature=0.0,
        top_p=0.9,
        top_k=20
    )['choices'][0]['text'].strip()
    return {
        'answer': rag_payload['answer'],
        'citations': rag_payload['citations'],
        'groundedness': groundedness_score,
        'relevance': relevance_score
    }
```

```python
evaluation_scores = {}
for key, question in business_queries.items():
    evaluation_scores[key] = generate_ground_relevance_response(question, active_retriever)

eval_records = []
for key, data in evaluation_scores.items():
    eval_records.append({
        'question': key,
        'groundedness': data['groundedness'],
        'relevance': data['relevance'],
        'citations': ', '.join(data['citations']),
        'answer_preview': data['answer'][:250] + ('...' if len(data['answer']) > 250 else '')
    })
evaluation_df = pd.DataFrame(eval_records)
display(evaluation_df)
```
